# URL Anomaly Detection

Use feature engineering and Isolation Forest to identify abnormal URL patterns.

```python
import pandas as pd
from sklearn.ensemble import IsolationForest

df = pd.DataFrame({
    "url": [
        "https://google.com",
        "https://amazon.com",
        "https://github.com",
        "http://secure-login-paypal-alert-123.com"
    ]
})

df["length"] = df["url"].str.len()
df["dots"] = df["url"].str.count(r"\.")
df["hyphens"] = df["url"].str.count("-")
df["digits"] = df["url"].str.count(r"\d")

X = df[["length", "dots", "hyphens", "digits"]]

model = IsolationForest(contamination=0.25, random_state=42)

df["result"] = model.fit_predict(X)
df["result"] = df["result"].map({1: "Normal", -1: "Anomalous"})

print(df)
```
